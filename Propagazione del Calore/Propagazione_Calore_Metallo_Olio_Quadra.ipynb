# Studio della Propagazione del Calore e Legge di Fourier (Barra Metallica con Olio e Sezione Quadrata)

Questo notebook analizza i dati sperimentali relativi alla propagazione del calore monodimensionale in una **barra metallica (ferro) cava riempita d'olio**, con sezione quadrata e una distanza tra i sensori di $dx = 10\text{ cm}$ ($0.1\text{ m}$). 
Vengono affrontati quattro aspetti principali:
1. **Derivazione teorica e bilancio energetico equivalente** per il sistema composto da metallo e olio, tenendo conto delle sezioni $A_m$, $A_o$, del perimetro esterno $P$, e calcolando la **diffusività efficace** $\alpha_{\mathrm{eff}}$, il coefficiente di perdita $\beta$ e il coefficiente di scambio equivalente $h$.
2. **Confronto dei parametri stimati** con valori di riferimento e con il modello equivalente metallo–olio.
3. **Fit analitico del riscaldamento** tramite il modello a ritardo di forma: 
   $$T(t) = T_{amb} + A \left(1 - e^{-t/\tau}\right)^b$$
   ed evidenziazione dell'analogia formale con la carica di un condensatore $RC$.
4. Derivazione teorica e fit della **fase di raffreddamento** (legge di raffreddamento di Newton).



## Dalla legge di Fourier all’equazione del calore

Consideriamo una barra di sezione costante $A$, con temperatura uniforme
in ogni sezione trasversale. La temperatura dipende dalla posizione lungo
la barra e dal tempo:

$$
T=T(x,t).
$$

Indichiamo con $k$ la conducibilità termica, con $\rho$ la densità,
con $C_s$ il calore specifico e con $P$ il perimetro esterno della sezione.
Supponiamo che queste proprietà siano costanti e che non vi siano sorgenti
interne di calore.

### La legge di Fourier

Il flusso termico per unità di superficie è

$$
\boxed{
j_Q(x,t)=-k\frac{\partial T}{\partial x}
}.
$$

Il segno meno indica che il calore fluisce dalle regioni a temperatura
maggiore verso quelle a temperatura minore.

La potenza termica che attraversa una sezione della barra è quindi

$$
\dot Q(x,t)=A\,j_Q(x,t).
$$

### Calore entrante e uscente da un tratto di barra

Consideriamo il tratto compreso tra $x$ e $x+dx$.

Nell’intervallo di tempo $dt$, il calore entrante è

$$
\delta Q_{\mathrm{in}}=A\,j_Q(x,t)\,dt,
$$

mentre quello uscente è

$$
\delta Q_{\mathrm{out}}=A\,j_Q(x+dx,t)\,dt.
$$

Sviluppando il flusso al primo ordine in $dx$:

$$
j_Q(x+dx,t)
\simeq
j_Q(x,t)+\frac{\partial j_Q}{\partial x}\,dx.
$$

Il calore netto ricevuto per conduzione è dunque

$$
\begin{aligned}
\delta Q_{\mathrm{cond}}
&=\delta Q_{\mathrm{in}}-\delta Q_{\mathrm{out}}\\
&=-A\frac{\partial j_Q}{\partial x}\,dx\,dt.
\end{aligned}
$$

**Osservazione fisica:** se entra più calore di quanto ne esca, il tratto
può aumentare la propria energia interna.

### Variazione dell’energia interna

Il tratto considerato ha volume $A\,dx$ e massa

$$
dm=\rho A\,dx.
$$

Durante il tempo $dt$, la variazione locale della temperatura è

$$
dT=\frac{\partial T}{\partial t}\,dt.
$$

La corrispondente variazione di energia interna è

$$
\boxed{
dU=\rho C_s A\,dx\,
\frac{\partial T}{\partial t}\,dt
}.
$$



### Dispersioni attraverso la superficie laterale

La superficie laterale del tratto è

$$
dS=P\,dx.
$$

Assumiamo uno scambio termico proporzionale alla differenza tra la
temperatura della barra e quella dell’ambiente:

$$
\delta Q_{\mathrm{perso}}
=hP\,dx\,(T-T_{\mathrm{amb}})\,dt,
$$

dove $h$ è il coefficiente di scambio termico, espresso in
$\mathrm{W/(m^2\,K)}$.

Se $T>T_{\mathrm{amb}}$, il termine rappresenta una perdita di energia.
Se $T<T_{\mathrm{amb}}$, descrive invece un trasferimento di calore
dall’ambiente alla barra.

Questa espressione assume uno scambio lineare. L’irraggiamento può essere
incluso in un coefficiente $h$ efficace soltanto mediante una opportuna
linearizzazione.

### Il bilancio energetico

Per conservazione dell’energia:

$$
\underbrace{dU}_{\text{energia accumulata}}
=
\underbrace{\delta Q_{\mathrm{cond}}}_{\text{conduzione netta}}
-
\underbrace{\delta Q_{\mathrm{perso}}}_{\text{dispersione laterale}}.
$$

Sostituendo:

$$
\rho C_s A\,dx\,
\frac{\partial T}{\partial t}\,dt
=
-A\frac{\partial j_Q}{\partial x}\,dx\,dt
-hP\,dx\,(T-T_{\mathrm{amb}})\,dt.
$$

Dividendo per $dx\,dt$:

$$
\rho C_s A\frac{\partial T}{\partial t}
=
-A\frac{\partial j_Q}{\partial x}
-hP(T-T_{\mathrm{amb}}).
$$

Dalla legge di Fourier:

$$
-\frac{\partial j_Q}{\partial x}
=
\frac{\partial}{\partial x}
\left(k\frac{\partial T}{\partial x}\right).
$$

Per conducibilità $k$ costante:

$$
-\frac{\partial j_Q}{\partial x}
=
k\frac{\partial^2 T}{\partial x^2}.
$$

Otteniamo quindi

$$
\boxed{
\underbrace{\rho C_s A\frac{\partial T}{\partial t}}
_{\text{accumulo per unità di lunghezza}}
=
\underbrace{kA\frac{\partial^2 T}{\partial x^2}}
_{\text{conduzione netta per unità di lunghezza}}
-
\underbrace{hP(T-T_{\mathrm{amb}})}
_{\text{perdita laterale per unità di lunghezza}}
}.
$$

Tutti i termini di questa equazione hanno unità di misura
$\mathrm{W/m}$.

### Diffusività termica e coefficiente di perdita

Dividendo per $\rho C_s A$, definiamo

$$
\boxed{
\alpha=\frac{k}{\rho C_s}
},
\qquad
[\alpha]=\mathrm{m^2/s},
$$

e

$$
\boxed{
\nu=\frac{hP}{\rho C_s A}
},
\qquad
[\nu]=\mathrm{s^{-1}}.
$$

L’equazione del calore con dispersioni laterali diventa

$$
\boxed{
\frac{\partial T}{\partial t}
=
\alpha\frac{\partial^2 T}{\partial x^2}
-\nu(T-T_{\mathrm{amb}})
}.
$$

Nel notebook, il coefficiente $\nu$ è indicato anche con il simbolo
$\beta$.

Se le dispersioni laterali sono trascurabili, ritroviamo la classica
equazione del calore:

$$
\boxed{
\frac{\partial T}{\partial t}
=
\alpha\frac{\partial^2 T}{\partial x^2}
}.
$$

### Significato delle capacità termiche

È importante distinguere:

$$
\rho C_s
\qquad
\left[\mathrm{\frac{J}{m^3\,K}}\right],
$$

che è la **capacità termica per unità di volume**, da

$$
\rho C_s A
\qquad
\left[\mathrm{\frac{J}{m\,K}}\right],
$$

che è la **capacità termica per unità di lunghezza**.

La capacità termica del piccolo tratto considerato è invece

$$
dC=\rho C_s A\,dx,
\qquad
[dC]=\mathrm{J/K}.
$$

###  Applicazione alla barra contenente olio

La derivazione precedente riguarda un mezzo descrivibile mediante
un’unica temperatura locale.

Per applicarla al sistema metallo–olio occorre assumere che, in ogni
sezione, parete e liquido siano quasi alla stessa temperatura e che il
trasporto dovuto al moto dell’olio sia trascurabile.

In queste condizioni, indicando con $A_m$ e $A_o$ le sezioni occupate
dal metallo e dall’olio, il bilancio equivalente è

$$
\left(\rho_m c_m A_m+\rho_o c_o A_o\right)
\frac{\partial T}{\partial t}
=
\left(k_m A_m+k_o A_o\right)
\frac{\partial^2 T}{\partial x^2}
-hP(T-T_{\mathrm{amb}}).
$$

La diffusività efficace risulta quindi

$$
\boxed{
\alpha_{\mathrm{eff}}
=
\frac{k_m A_m+k_o A_o}
{\rho_m c_m A_m+\rho_o c_o A_o}
}.
$$

Questa espressione dipende dalle proprietà e dalle sezioni dei due
materiali: non è una semplice media delle loro diffusività.

**Domanda di riflessione:** se due punti vicini della barra hanno
temperature diverse, basta questa informazione per sapere se un
determinato tratto si sta riscaldando? Oppure occorre confrontare il
calore entrante, quello uscente e quello disperso?


```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter
from scipy.optimize import curve_fit

# Impostazioni grafiche globali
plt.rcParams.update({
    'figure.figsize': (10, 5),
    'axes.grid': True,
    'grid.alpha': 0.3,
    'font.size': 11
})

# --- PARAMETRI GEOMETRICI E FISICI DEL SISTEMA METALLO–OLIO (SEZIONE QUADRATA) ---
dx = 0.10          # Distanza tra i sensori in metri (10 cm)
L_ext = 0.025      # ASSUNTO: lato esterno della barra quadrata (2.5 cm)
spessore = 0.0025  # ASSUNTO: spessore della parete metallica (2.5 mm)
L_int = L_ext - 2 * spessore  # Lato cavità interna occupata dall'olio

# Aree delle sezioni
A_m = L_ext**2 - L_int**2     # Sezione metallica (Ferro)
A_o = L_int**2                # Sezione interna (Olio)
P = 4 * L_ext                 # Perimetro esterno esposto all'aria

# Proprietà fisiche del Ferro (Metallo)
rho_m = 7874.0  # Densità del ferro [kg/m^3]
c_m = 450.0     # Calore specifico del ferro [J/(kg K)]
k_m = 80.0      # Conducibilità termica stimata del ferro [W/(m K)]

# Proprietà fisiche dell'Olio
rho_o = 880.0   # Densità dell'olio [kg/m^3]
c_o = 1800.0    # Calore specifico dell'olio [J/(kg K)]
k_o = 0.15      # Conducibilità termica dell'olio [W/(m K)]

# Capacità termiche per unità di lunghezza [J/(m K)] e diffusività teorica
Cap_termica_m = rho_m * c_m * A_m
Cap_termica_o = rho_o * c_o * A_o
alpha_eff_teorica = (k_m * A_m + k_o * A_o) / (Cap_termica_m + Cap_termica_o)

print(f"Sezione metallo (Am): {A_m:.5f} m^2 | Sezione olio (Ao): {A_o:.5f} m^2")
print(f"Diffusività efficace teorica attesa (alpha_eff): {alpha_eff_teorica:.3e} m^2/s")

# Caricamento dei dati sperimentali
path_main = "Legge_Fourier.csv"

try:
    df_main = pd.read_csv(path_main)
    t_sec = df_main.iloc[:, 0].to_numpy() * 60  # Conversione min -> s
    T_matrix = df_main.iloc[:, 1:5].to_numpy()  # Sensori T1, T2, T3, T4
except Exception as e:
    print("File locale non trovato. Generazione dati simulati di test...")
    t_sec = np.linspace(0, 3600, 600)
    T1 = 20.5 + 45.0 * (1 - np.exp(-t_sec / 550))**1.4 * np.exp(-np.maximum(0, t_sec - 1800) / 1200)
    T2 = 20.5 + 38.0 * (1 - np.exp(-t_sec / 600))**1.5 * np.exp(-np.maximum(0, t_sec - 1800) / 1250)
    T3 = 20.5 + 31.0 * (1 - np.exp(-t_sec / 650))**1.6 * np.exp(-np.maximum(0, t_sec - 1800) / 1300)
    T4 = 20.5 + 24.0 * (1 - np.exp(-t_sec / 700))**1.7 * np.exp(-np.maximum(0, t_sec - 1800) / 1350)
    T_matrix = np.column_stack([T1, T2, T3, T4])

T_amb = float(T_matrix[t_sec <= 60].mean())
print(f"Temperatura ambiente stimata (T_amb): {T_amb:.2f} °C")
```

```python
# Grafico 1: Profilo temporale delle temperature
for i in range(T_matrix.shape[1]):
    plt.plot(t_sec / 60, T_matrix[:, i], label=f'Sensore T{i+1}', linewidth=1.8)

plt.title('Evoluzione Temporale della Temperatura', fontsize=12, fontweight='bold')
plt.xlabel('Tempo (min)')
plt.ylabel('Temperatura (°C)')
plt.legend(loc='upper right')
```

## 2. Stima Sperimentale dei Parametri ($\alpha_{\mathrm{eff}}$, $\beta$, $h$)

Riscrivendo l'equazione di bilancio con i sensori distanziati di $dx = 0.1\text{m}$, applichiamo un fit multivariato per estrarre $\alpha_{\mathrm{eff}}$ e $\beta$:


La distanza tra i sensori, $dx=0.10\,\mathrm m$, è misurata. Il lato esterno di 2,5 cm e lo spessore di 2,5 mm sono invece **assunti**: aree e perimetro derivati dipendono da queste ipotesi. Le proprietà di ferro e olio inserite nel codice sono valori di riferimento trattati come costanti. Il confronto teorico non è una misura indipendente del materiale.



$$\frac{\partial T}{\partial t} = \alpha_{\mathrm{eff}} \left(\frac{\partial^2 T}{\partial x^2}\right) - \beta (T - T_{\mathrm{amb}})$$


Dal coefficiente di perdita $\beta$, ricaviamo il **coefficiente di scambio equivalente** $h$:

$$h = \frac{\beta \left(\rho_m c_m A_m + \rho_o c_o A_o\right)}{P}.$$

Il valore di $h$ è condizionato dalla geometria assunta e dalle proprietà dei materiali inserite. Può includere convezione, irraggiamento linearizzato e altri scambi; non rappresenta necessariamente il solo coefficiente convettivo.


Per verificare l'equazione di Fourier, calcoliamo:
1. La derivata temporale $\frac{\partial T}{\partial t}$ applicando il filtro di Savitzky-Golay sui dati temporali per ridurre il rumore.
2. La derivata spaziale seconda $\frac{\partial^2 T}{\partial x^2}$ con lo schema alle differenze finite centrate:
$$\frac{\partial^2 T_i}{\partial x^2} \approx \frac{T_{i+1} - 2T_i + T_{i-1}}{\Delta x^2}$$

### Incertezze: stima semplice dalla regressione

Ponendo $\mathbf y=\partial T/\partial t$ e $\mathbf p=(\alpha,\beta)$, scriviamo $\mathbf y=X\mathbf p$, con colonne $\partial^2T/\partial x^2$ e $-(T-T_{\mathrm{amb}})$. Dai residui $\mathbf r=\mathbf y-X\hat{\mathbf p}$ calcoliamo:

$$s^2=\frac{\sum_j r_j^2}{n-2},\qquad
\operatorname{Cov}(\hat{\mathbf p})=s^2(X^{\mathsf T}X)^{-1}.$$

Le radici degli elementi diagonali sono gli errori standard $u_\alpha$ e $u_\beta$, riportati come **incertezze formali a un errore standard**. È il metodo ordinario più semplice: assume regressori esatti e residui indipendenti con varianza costante. Qui le derivate sono rumorose e i residui possono essere correlati: questi errori **non sono l'incertezza sperimentale completa e possono sottostimarla**. Non comprendono distanze, temperatura ambiente, geometria, proprietà dei materiali o errore del modello.


```python
dt = np.median(np.diff(t_sec))
window_length = 15 if len(t_sec) >= 15 else 5
poly_order = 2

# Derivata temporale per il sensore centrale T2
dT2_dt = savgol_filter(T_matrix[:, 1], window_length, poly_order, deriv=1, delta=dt)

# Derivata spaziale seconda con dx = 10 cm (0.1 m)
d2T_dx2_T2 = (T_matrix[:, 2] - 2*T_matrix[:, 1] + T_matrix[:, 0]) / (dx**2)
delta_T_T2 = T_matrix[:, 1] - T_amb

# Regressione lineare multivariata: dT/dt = alpha_eff * (d2T/dx2) - beta * (T - T_amb)
X = np.column_stack([d2T_dx2_T2, -delta_T_T2])
params, residuals, rank, s = np.linalg.lstsq(X, dT2_dt, rcond=None)

alpha_eff_fit, beta_fit = params[0], params[1]

# Errori standard formali della regressione ordinaria (un errore standard)
residui_fit = dT2_dt - X @ params
gradi_liberta = len(dT2_dt) - len(params)
if rank < len(params) or gradi_liberta <= 0:
    raise ValueError("Dati insufficienti o regressione non identificabile.")
varianza_residui = np.sum(residui_fit**2) / gradi_liberta
cov_params = varianza_residui * np.linalg.inv(X.T @ X)
u_alpha, u_beta = np.sqrt(np.diag(cov_params))

# Coefficiente di scambio equivalente: geometria e proprietà assunte
h_fit = beta_fit * (Cap_termica_m + Cap_termica_o) / P

print("--- STIMA SPERIMENTALE DEI PARAMETRI (SISTEMA METALLO-OLIO) ---")
print(f"Diffusività efficace (alpha_eff): ({alpha_eff_fit:.3e} ± {u_alpha:.2e}) m^2/s")
print(f"Coefficiente di perdita beta: ({beta_fit:.3e} ± {u_beta:.2e}) s^-1")
print(f"Coefficiente di scambio equivalente h: {h_fit:.2f} W/(m^2 K)")
print("± = errore standard formale OLS: non include tutte le incertezze sperimentali.")
```

## 3. Confronto con i valori di riferimento e il modello equivalente

| Materiale / Sistema | Conducibilità $k$ [W/(m K)] | Calore specifico $C_s$ [J/(kg K)] | Densità $\rho$ [kg/m³] | Diffusività $\alpha$ [m²/s] |
| :--- | :---: | :---: | :---: | :---: |
| **Ferro (parete; riferimento assunto)** | 80.0 | 450 | 7874 | $2.258\times10^{-5}$ |
| **Olio (interno; riferimento assunto)** | 0.15 | 1800 | 880 | $9.470\times10^{-8}$ |
| **Sistema composto (modello equivalente)** | — | — | — | **$1.262\times10^{-5}$** |

Le prime due diffusività sono calcolate con $\alpha=k/(\rho C_s)$ dai valori di riferimento inseriti nel codice. Quella del composto è calcolata con

$$\alpha_{\mathrm{eff}}=\frac{k_m A_m+k_o A_o}{\rho_m c_m A_m+\rho_o c_o A_o},$$

usando lato e spessore **assunti**. Non è un valore tabulato del composto né una semplice media delle diffusività. Se si modificano i parametri, il valore teorico aggiornato è quello stampato dalla cella dei parametri.

**Commento:** rispetto alla stessa parete cava priva di olio, l'olio aumenta la capacità termica per unità di lunghezza; con i parametri assunti riduce la diffusività efficace. Il modello richiede temperatura quasi uniforme nella sezione e trasporto convettivo interno trascurabile.


## 4. Fit Analitico del Riscaldamento con Ritardo di Forma

### Modello a Ritardo di Forma
Il profilo temporale del riscaldamento è descritto dalla funzione a ritardo di forma (analogia con la carica del condensatore $RC$):

$$\boxed{T(t) = T_{amb} + A \left(1 - e^{-t/\tau}\right)^b}$$

dove la costante di tempo termica $\tau = R_{th} C_{th}$ tiene conto della capacità termica equivalente del sistema composito metallo-olio.

```python
# Identificazione della fase di riscaldamento per T2
T2 = T_matrix[:, 0]
idx_peak = np.argmax(T2)

t_heat = t_sec[:idx_peak]
T_heat = T2[:idx_peak]

def model_heating_delay(t, A, tau, b):
    return T_amb + A * (1.0 - np.exp(-t / tau))**b

p0_heat = [T_heat.max() - T_amb, 500.0, 1.4]
popt_heat, pcov_heat = curve_fit(model_heating_delay, t_heat, T_heat, p0=p0_heat)
A_fit, tau_heat_fit, b_fit = popt_heat

print("--- RISULTATI FIT RISCALDAMENTO CON RITARDO DI FORMA ---")
print(f"Ampiezza A (T_inf - T_amb): {A_fit:.2f} °C")
print(f"Costante di tempo (tau): {tau_heat_fit:.1f} s ({tau_heat_fit/60:.2f} min)")
print(f"Esponente di forma (b): {b_fit:.3f}")
```

## 5. Analisi del Raffreddamento (Decadimento Newtoniano)

### Derivazione Teorica
Durante il raffreddamento, il bilancio energetico si riduce alla dispersione convettiva/radiativa verso l'ambiente:

$$\left(\rho_m c_m A_m + \rho_o c_o A_o\right) \frac{dT}{dt} = -h P (T - T_{amb})$$

Integrando nel tempo, la legge di decadimento assume la forma esponenziale di Newton:

$$\boxed{T(t) = T_{amb} + (T_{init} - T_{amb}) e^{-\frac{t - t_0}{\tau_{cool}}}}$$

dove la costante di tempo del raffreddamento è $\tau_{cool} = \frac{\rho_m c_m A_m + \rho_o c_o A_o}{h P} = \frac{1}{\beta}$.

```python
t_cool = t_sec[idx_peak:] - t_sec[idx_peak]
T_cool = T2[idx_peak:]

def model_cooling(t, T_init, tau_cool):
    return T_amb + (T_init - T_amb) * np.exp(-t / tau_cool)

p0_cool = [T_cool[0], 700.0]
popt_cool, pcov_cool = curve_fit(model_cooling, t_cool, T_cool, p0=p0_cool)
T_init_fit, tau_cool_fit = popt_cool

print("--- RISULTATI FIT RAFFREDDAMENTO ---")
print(f"Temperatura iniziale (T_init): {T_init_fit:.2f} °C")
print(f"Costante di tempo raffreddamento (tau_cool): {tau_cool_fit:.1f} s ({tau_cool_fit/60:.2f} min)")

# --- VISUALIZZAZIONE COMPARATIVA --- 
fig, axs = plt.subplots(1, 2, figsize=(14, 5))

# Plot Riscaldamento
axs[0].plot(t_heat / 60, T_heat, label='Dati Sperimentali (T2)', color='orangered', lw=1.8)
axs[0].plot(t_heat / 60, model_heating_delay(t_heat, *popt_heat), 
             label=f'Fit Ritardo Forma ($\tau$={tau_heat_fit:.0f}s, b={b_fit:.2f})', 
             color='black', ls='--', lw=2)
axs[0].set_title('Riscaldamento (Sistema Metallo-Olio)', fontweight='bold')
axs[0].set_xlabel('Tempo (min)')
axs[0].set_ylabel('Temperatura (°C)')
axs[0].legend()

# Plot Raffreddamento
axs[1].plot(t_cool / 60, T_cool, label='Dati Sperimentali (T2)', color='dodgerblue', lw=1.8)
axs[1].plot(t_cool / 60, model_cooling(t_cool, *popt_cool), 
             label=f'Fit Newton ($\tau_{{cool}}$={tau_cool_fit:.0f}s)', 
             color='black', ls='--', lw=2)
axs[1].axhline(T_amb, color='gray', ls=':', label=f'T_amb ({T_amb:.1f}°C)')
axs[1].set_title('Raffreddamento Newtoniano (Metallo-Olio)', fontweight='bold')
axs[1].set_xlabel('Tempo dall\'inizio del raffreddamento (min)')
axs[1].set_ylabel('Temperatura (°C)')
axs[1].legend()

plt.tight_layout()
plt.show()
```
