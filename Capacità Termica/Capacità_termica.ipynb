C_0=c_{acqua}m_{acqua} \frac{T_{acq} - T_{eq}}{T_{eq}-T_0}
$$

## Propagazione degli errori sulla capacità termica


La capacità termica del calorimetro è data dalla relazione:

$$
C_0 = c_{\text{acqua}} m_{\text{acqua}}
\frac{T_{\text{acq}} - T_{\text{eq}}}{T_{\text{eq}} - T_0}
$$

Assumendo che il calore specifico dell'acqua $c_{\text{acqua}}$
sia noto senza incertezza, utilizziamo la propagazione lineare
degli errori:

$$
\Delta C_0 \simeq
\left|\frac{\partial C_0}{\partial m_{\text{acqua}}}\right|
\Delta m_{\text{acqua}}
+
\left|\frac{\partial C_0}{\partial T_{\text{acq}}}\right|
\Delta T_{\text{acq}}
+
\left|\frac{\partial C_0}{\partial T_{\text{eq}}}\right|
\Delta T_{\text{eq}}
+
\left|\frac{\partial C_0}{\partial T_0}\right|
\Delta T_0
$$

### Calcolo delle derivate parziali

**Rispetto alla massa dell'acqua:**

$$
\frac{\partial C_0}{\partial m_{\text{acqua}}}
=
c_{\text{acqua}}
\frac{T_{\text{acq}}-T_{\text{eq}}}{T_{\text{eq}}-T_0}
$$

**Rispetto alla temperatura iniziale dell'acqua:**

$$
\frac{\partial C_0}{\partial T_{\text{acq}}}
=
\frac{c_{\text{acqua}}m_{\text{acqua}}}
{T_{\text{eq}}-T_0}
$$

**Rispetto alla temperatura di equilibrio:**

$$
\frac{\partial C_0}{\partial T_{\text{eq}}}
=
-c_{\text{acqua}}m_{\text{acqua}}
\frac{T_{\text{acq}}-T_0}{(T_{\text{eq}}-T_0)^2}
$$

**Rispetto alla temperatura iniziale del calorimetro:**

$$
\frac{\partial C_0}{\partial T_0}
=
c_{\text{acqua}}m_{\text{acqua}}
\frac{T_{\text{acq}}-T_{\text{eq}}}
{(T_{\text{eq}}-T_0)^2}
$$

### Incertezza assoluta

Sostituendo le derivate parziali, otteniamo:

$$
\boxed{
\begin{aligned}
\Delta C_0 \simeq c_{\text{acqua}}\Bigg[
&\left|
\frac{T_{\text{acq}}-T_{\text{eq}}}
{T_{\text{eq}}-T_0}
\right|\Delta m_{\text{acqua}}
\\[6pt]
+&\left|
\frac{m_{\text{acqua}}}
{T_{\text{eq}}-T_0}
\right|\Delta T_{\text{acq}}
\\[6pt]
+&\left|
\frac{m_{\text{acqua}}(T_{\text{acq}}-T_0)}
{(T_{\text{eq}}-T_0)^2}
\right|\Delta T_{\text{eq}}
\\[6pt]
+&\left|
\frac{m_{\text{acqua}}(T_{\text{acq}}-T_{\text{eq}})}
{(T_{\text{eq}}-T_0)^2}
\right|\Delta T_0
\Bigg]
\end{aligned}
}
$$

### Incertezza relativa

Dividendo entrambi i membri per il valore assoluto
della capacità termica, otteniamo:

$$
\boxed{
\begin{aligned}
\frac{\Delta C_0}{|C_0|} \simeq
&\frac{\Delta m_{\text{acqua}}}{m_{\text{acqua}}}
+
\frac{\Delta T_{\text{acq}}}
{|T_{\text{acq}}-T_{\text{eq}}|}
\\[6pt]
&+
\left|
\frac{T_{\text{acq}}-T_0}
{(T_{\text{acq}}-T_{\text{eq}})
(T_{\text{eq}}-T_0)}
\right|\Delta T_{\text{eq}}
\\[6pt]
&+
\frac{\Delta T_0}{|T_{\text{eq}}-T_0|}
\end{aligned}
}
$$

### Caso particolare: stessa incertezza sulle temperature

Se tutte le temperature vengono misurate con lo stesso
strumento e assumiamo:

$$
\Delta T_{\text{acq}} =
\Delta T_{\text{eq}} =
\Delta T_0 = \Delta T
$$

allora:

$$
\boxed{
\begin{aligned}
\Delta C_0 \simeq c_{\text{acqua}}\Bigg[
&\left|
\frac{T_{\text{acq}}-T_{\text{eq}}}
{T_{\text{eq}}-T_0}
\right|\Delta m_{\text{acqua}}
\\[6pt]
&+m_{\text{acqua}}\Delta T
\left(
\frac{1}{|T_{\text{eq}}-T_0|}
+
\frac{|T_{\text{acq}}-T_0|}
{(T_{\text{eq}}-T_0)^2}
+
\frac{|T_{\text{acq}}-T_{\text{eq}}|}
{(T_{\text{eq}}-T_0)^2}
\right)
\Bigg]
\end{aligned}
}
$$

## Calcolo del calore specifico del calorimetro



Una volta determinata la capacità termica $C_0$, possiamo
calcolare il calore specifico del materiale di cui è costituito
il calorimetro.

Il calore specifico è definito come il rapporto tra la
capacità termica e la massa del corpo:

$$
\boxed{
c_0 = \frac{C_0}{m_0}
}
$$

dove:
- $C_0$ è la capacità termica del calorimetro;
- $m_0$ è la massa del calorimetro.

Sostituendo l'espressione di $C_0$, otteniamo:

$$
\boxed{
c_0 =
\frac{c_{\text{acqua}}m_{\text{acqua}}}{m_0}
\frac{T_{\text{acq}}-T_{\text{eq}}}{T_{\text{eq}}-T_0}
}
$$

L'unità di misura del calore specifico è:

$$
[c_0] = \frac{\text{J}}{\text{g}\cdot{}^\circ\text{C}}
$$

### Propagazione degli errori

Consideriamo la relazione:

$$
c_0 = \frac{C_0}{m_0}
$$

Applicando la propagazione lineare degli errori:

$$
\Delta c_0 \simeq
\left|\frac{\partial c_0}{\partial C_0}\right|
\Delta C_0
+
\left|\frac{\partial c_0}{\partial m_0}\right|
\Delta m_0
$$

Calcoliamo le derivate parziali.

**Rispetto alla capacità termica:**

$$
\frac{\partial c_0}{\partial C_0} = \frac{1}{m_0}
$$

**Rispetto alla massa del calorimetro:**

$$
\frac{\partial c_0}{\partial m_0} =
-\frac{C_0}{m_0^2}
$$

Sostituendo nella formula di propagazione:

$$
\boxed{
\Delta c_0 \simeq
\frac{\Delta C_0}{m_0}
+
\frac{|C_0|}{m_0^2}\Delta m_0
}
$$

### Incertezza relativa

Dividendo entrambi i membri per il valore assoluto
del calore specifico:

$$
\boxed{
\frac{\Delta c_0}{|c_0|}
\simeq
\frac{\Delta C_0}{|C_0|}
+
\frac{\Delta m_0}{m_0}
}
$$

L'incertezza relativa percentuale è:

$$
\boxed{
\varepsilon_{c_0} =
\frac{\Delta c_0}{|c_0|}\cdot 100
}
$$

### Risultato della misura

Il risultato sperimentale deve essere espresso nella forma:

$$
\boxed{
c_0 = (\overline{c}_0 \pm \Delta c_0)
\ \frac{\text{J}}{\text{g}\cdot{}^\circ\text{C}}
}
$$

```python
# ==============================
# PROPAGAZIONE LINEARE DEGLI ERRORI
# ==============================

# Derivate parziali in valore assoluto

dC_dm = abs(c_acqua * (Tacq - Teq) / (Teq - T0))

dC_dTacq = abs(c_acqua * m_acqua / (Teq - T0))

dC_dTeq = abs(
    -c_acqua * m_acqua * (Tacq - T0) / (Teq - T0)**2
)

dC_dT0 = abs(
    c_acqua * m_acqua * (Tacq - Teq) / (Teq - T0)**2
)

# Incertezza assoluta

dC0 = (
    dC_dm * dm_acqua
    + dC_dTacq * dTacq
    + dC_dTeq * dTeq
    + dC_dT0 * dT0
)

# Incertezza relativa percentuale

errore_percentuale = (dC0 / abs(C0)) * 100

# ==============================
# RISULTATI
# ==============================

print(f"Capacità termica: C0 = ({C0:.2f} ± {dC0:.2f}) J/°C")
print(f"calore specifico calorimetro: c0 = ({C0/m_0:.2f}) J/(g°C)")
print(f"Incertezza su C0: {dC0:.2f} J/°C")
print(f"Incertezza relativa: {errore_percentuale:.2f} %")
```

# Parte II: Stima della Capacità Termica di un materiale ignoto

$$
C_X=(C_{eq}+C_0)\frac{T_0 - T_{eq}}{T_{eq}-T_{x}}
$$

## Parte II: Stima della capacità termica di un materiale ignoto

La capacità termica di un corpo ignoto è data dalla relazione:

$$
C_X=(C_{\text{eq}}+C_0)
\frac{T_0-T_{\text{eq}}}{T_{\text{eq}}-T_X}
$$

dove la capacità termica dell'acqua è:

$$
C_{\text{eq}}=c_{\text{acqua}}m_{\text{acqua}}
$$

### Propagazione degli errori

Utilizziamo la propagazione lineare delle incertezze:

$$
\begin{aligned}
\Delta C_X \simeq&
\left|\frac{\partial C_X}{\partial C_{\text{eq}}}\right|
\Delta C_{\text{eq}}
+
\left|\frac{\partial C_X}{\partial C_0}\right|
\Delta C_0
\\[6pt]
&+
\left|\frac{\partial C_X}{\partial T_0}\right|
\Delta T_0
+
\left|\frac{\partial C_X}{\partial T_{\text{eq}}}\right|
\Delta T_{\text{eq}}
\\[6pt]
&+
\left|\frac{\partial C_X}{\partial T_X}\right|
\Delta T_X
\end{aligned}
$$

### Calcolo delle derivate parziali

**Rispetto alla capacità termica dell'acqua:**

$$
\frac{\partial C_X}{\partial C_{\text{eq}}}
=
\frac{T_0-T_{\text{eq}}}{T_{\text{eq}}-T_X}
$$

**Rispetto alla capacità termica del calorimetro:**

$$
\frac{\partial C_X}{\partial C_0}
=
\frac{T_0-T_{\text{eq}}}{T_{\text{eq}}-T_X}
$$

**Rispetto alla temperatura iniziale:**

$$
\frac{\partial C_X}{\partial T_0}
=
\frac{C_{\text{eq}}+C_0}{T_{\text{eq}}-T_X}
$$

**Rispetto alla temperatura di equilibrio:**

$$
\frac{\partial C_X}{\partial T_{\text{eq}}}
=
(C_{\text{eq}}+C_0)
\frac{T_X-T_0}{(T_{\text{eq}}-T_X)^2}
$$

**Rispetto alla temperatura iniziale del corpo:**

$$
\frac{\partial C_X}{\partial T_X}
=
(C_{\text{eq}}+C_0)
\frac{T_0-T_{\text{eq}}}{(T_{\text{eq}}-T_X)^2}
$$

### Incertezza assoluta

Sostituendo le derivate parziali:

$$
\boxed{
\begin{aligned}
\Delta C_X \simeq&
\left|\frac{T_0-T_{\text{eq}}}{T_{\text{eq}}-T_X}\right|
(\Delta C_{\text{eq}}+\Delta C_0)
\\[6pt]
&+
\left|\frac{C_{\text{eq}}+C_0}{T_{\text{eq}}-T_X}\right|
\Delta T_0
\\[6pt]
&+
\left|
\frac{(C_{\text{eq}}+C_0)(T_X-T_0)}
{(T_{\text{eq}}-T_X)^2}
\right|\Delta T_{\text{eq}}
\\[6pt]
&+
\left|
\frac{(C_{\text{eq}}+C_0)(T_0-T_{\text{eq}})}
{(T_{\text{eq}}-T_X)^2}
\right|\Delta T_X
\end{aligned}
}
$$

Poiché:

$$
C_{\text{eq}}=c_{\text{acqua}}m_{\text{acqua}}
$$

assumendo noto il calore specifico dell'acqua:

$$
\Delta C_{\text{eq}}=
c_{\text{acqua}}\Delta m_{\text{acqua}}
$$

## Calcolo del calore specifico del materiale ignoto

Il calore specifico è dato da:

$$
\boxed{
c_X=\frac{C_X}{m_X}
}
$$

### Propagazione degli errori

$$
\Delta c_X\simeq
\left|\frac{\partial c_X}{\partial C_X}\right|
\Delta C_X
+
\left|\frac{\partial c_X}{\partial m_X}\right|
\Delta m_X
$$

Le derivate parziali sono:

$$
\frac{\partial c_X}{\partial C_X}=\frac{1}{m_X}
$$

$$
\frac{\partial c_X}{\partial m_X}=-\frac{C_X}{m_X^2}
$$

Pertanto:

$$
\boxed{
\Delta c_X\simeq
\frac{\Delta C_X}{m_X}
+
\frac{|C_X|}{m_X^2}\Delta m_X
}
$$

### Incertezza relativa

$$
\boxed{
\frac{\Delta c_X}{|c_X|}
\simeq
\frac{\Delta C_X}{|C_X|}
+
\frac{\Delta m_X}{m_X}
}
$$

### Conversione in calorie

$$
c_X^{(\text{cal})}=\frac{c_X}{4.184}
$$

$$
\Delta c_X^{(\text{cal})}=\frac{\Delta c_X}{4.184}
$$

### Risultato della misura

$$
\boxed{
c_X=(\overline{c}_X\pm\Delta c_X)
\frac{\text{cal}}{\text{g}\cdot{}^\circ\text{C}}
}
$$

```python
# ============================================
# PARTE II
# CAPACITÀ TERMICA DI UN MATERIALE IGNOTO (RAME)
# ============================================

# DATI SPERIMENTALI

T0_X =23.9      # Temperatura iniziale acqua e calorimetro (°C)
TX = 76.8        # Temperatura iniziale corpo ignoto (°C)
Teq_X = 29.5     # Temperatura di equilibrio (°C)

m_acqua_con_beker = 1019.9 # Massa dell'acqua (g)
m_beker=248.0       # Massa del becker (g)
m_acqua_X = m_acqua_con_beker - m_beker  # Massa dell'acqua (g)
m_X = 1008.2       # Massa del corpo ignoto (g)

# Valori già determinati nella Parte I:
# C0 = ...
# dC0 = ...
# c_acqua = 4.186 J/(g°C)


# ============================================
# INCERTEZZE SPERIMENTALI
# ============================================

dT0_X = 0.1
dTX = 0.1
dTeq_X = 0.1

dm_acqua_X = 0.2
dm_X = 0.1

# ============================================
# CAPACITÀ TERMICA DELL'ACQUA
# ============================================

Ceq = c_acqua * m_acqua_X

# Incertezza sulla capacità termica dell'acqua

dCeq = c_acqua * dm_acqua_X


# ============================================
# CAPACITÀ TERMICA DEL CORPO IGNOTO
# ============================================

CX = (Ceq + C0) * (T0_X - Teq_X) / (Teq_X - TX)


# ============================================
# PROPAGAZIONE LINEARE DEGLI ERRORI
# ============================================

# Derivata rispetto a Ceq

dCX_dCeq = abs(
    (T0_X - Teq_X) / (Teq_X - TX)
)

# Derivata rispetto a C0

dCX_dC0 = abs(
    (T0_X - Teq_X) / (Teq_X - TX)
)

# Derivata rispetto a T0

dCX_dT0 = abs(
    (Ceq + C0) / (Teq_X - TX)
)

# Derivata rispetto a Teq

dCX_dTeq = abs(
    (Ceq + C0) * (TX - T0_X) / (Teq_X - TX)**2
)

# Derivata rispetto a TX

dCX_dTX = abs(
    (Ceq + C0) * (T0_X - Teq_X) / (Teq_X - TX)**2
)


# ============================================
# INCERTEZZA ASSOLUTA SU CX
# ============================================

dCX = (
    dCX_dCeq * dCeq
    + dCX_dC0 * dC0
    + dCX_dT0 * dT0_X
    + dCX_dTeq * dTeq_X
    + dCX_dTX * dTX
)

# Incertezza relativa percentuale

err_CX = dCX / abs(CX) * 100


# ============================================
# CALORE SPECIFICO DEL CORPO IGNOTO
# ============================================

cX = CX / m_X

# Propagazione lineare degli errori

dcX = (
    dCX / m_X
    + abs(CX) / m_X**2 * dm_X
)

# Incertezza relativa percentuale

err_cX = dcX / abs(cX) * 100


# ============================================
# CONVERSIONE IN CALORIE
# ============================================

cX_cal = cX / 4.184

dcX_cal = dcX / 4.184


# ============================================
# RISULTATI
# ============================================

print("====================================")
print("CAPACITÀ TERMICA DEL CORPO IGNOTO")
print("====================================")

print(f"CX = ({CX:.2f} ± {dCX:.2f}) J/°C")
print(f"Incertezza relativa = {err_CX:.2f} %")

print("\n====================================")
print("CALORE SPECIFICO DEL CORPO IGNOTO")
print("====================================")

print(f"cX = ({cX:.3f} ± {dcX:.3f}) J/(g°C)")
print(f"Incertezza relativa = {err_cX:.2f} %")

print("\n====================================")
print("CALORE SPECIFICO IN CALORIE")
print("====================================")

print(f"cX = ({cX_cal:.4f} ± {dcX_cal:.4f}) cal/(g°C)")
```

```python
# ============================================
# PARTE II
# CAPACITÀ TERMICA DI UN MATERIALE IGNOTO (Ottone)
# ============================================

# DATI SPERIMENTALI

T0_X =29.3 #23.9      # Temperatura iniziale acqua e calorimetro (°C)
TX = 77.7#76.8        # Temperatura iniziale corpo ignoto (°C)
Teq_X = 34.7 #29.5     # Temperatura di equilibrio (°C)

m_acqua_con_beker = 1019.9 # Massa dell'acqua (g)
m_beker=248.0       # Massa del becker (g)
m_acqua_X = m_acqua_con_beker - m_beker  # Massa dell'acqua (g)
m_X = 1007#1008.2       # Massa del corpo ignoto (g)

# Valori già determinati nella Parte I:
# C0 = ...
# dC0 = ...
# c_acqua = 4.186 J/(g°C)


# ============================================
# INCERTEZZE SPERIMENTALI
# ============================================

dT0_X = 0.1
dTX = 0.1
dTeq_X = 0.1

dm_acqua_X = 0.1
dm_X = 0.1

# ============================================
# CAPACITÀ TERMICA DELL'ACQUA
# ============================================

Ceq = c_acqua * m_acqua_X

# Incertezza sulla capacità termica dell'acqua

dCeq = c_acqua * dm_acqua_X


# ============================================
# CAPACITÀ TERMICA DEL CORPO IGNOTO
# ============================================

CX = (Ceq + C0) * (T0_X - Teq_X) / (Teq_X - TX)


# ============================================
# PROPAGAZIONE LINEARE DEGLI ERRORI
# ============================================

# Derivata rispetto a Ceq

dCX_dCeq = abs(
    (T0_X - Teq_X) / (Teq_X - TX)
)

# Derivata rispetto a C0

dCX_dC0 = abs(
    (T0_X - Teq_X) / (Teq_X - TX)
)

# Derivata rispetto a T0

dCX_dT0 = abs(
    (Ceq + C0) / (Teq_X - TX)
)

# Derivata rispetto a Teq

dCX_dTeq = abs(
    (Ceq + C0) * (TX - T0_X) / (Teq_X - TX)**2
)

# Derivata rispetto a TX

dCX_dTX = abs(
    (Ceq + C0) * (T0_X - Teq_X) / (Teq_X - TX)**2
)


# ============================================
# INCERTEZZA ASSOLUTA SU CX
# ============================================

dCX = (
    dCX_dCeq * dCeq
    + dCX_dC0 * dC0
    + dCX_dT0 * dT0_X
    + dCX_dTeq * dTeq_X
    + dCX_dTX * dTX
)

# Incertezza relativa percentuale

err_CX = dCX / abs(CX) * 100


# ============================================
# CALORE SPECIFICO DEL CORPO IGNOTO
# ============================================

cX = CX / m_X

# Propagazione lineare degli errori

dcX = (
    dCX / m_X
    + abs(CX) / m_X**2 * dm_X
)

# Incertezza relativa percentuale

err_cX = dcX / abs(cX) * 100


# ============================================
# CONVERSIONE IN CALORIE
# ============================================

cX_cal = cX / 4.184

dcX_cal = dcX / 4.184


# ============================================
# RISULTATI
# ============================================

print("====================================")
print("CAPACITÀ TERMICA DEL CORPO IGNOTO")
print("====================================")

print(f"CX = ({CX:.2f} ± {dCX:.2f}) J/°C")
print(f"Incertezza relativa = {err_CX:.2f} %")

print("\n====================================")
print("CALORE SPECIFICO DEL CORPO IGNOTO")
print("====================================")

print(f"cX = ({cX:.3f} ± {dcX:.3f}) J/(g°C)")
print(f"Incertezza relativa = {err_cX:.2f} %")

print("\n====================================")
print("CALORE SPECIFICO IN CALORIE")
print("====================================")

print(f"cX = ({cX_cal:.4f} ± {dcX_cal:.4f}) cal/(g°C)")
```
