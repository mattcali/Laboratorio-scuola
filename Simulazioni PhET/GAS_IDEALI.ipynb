# GAS IDEALI

Utilizzando il simulatore https://phet.colorado.edu/sims/html/gas-properties/latest/gas-properties_all.html studiare le proprietà dei gas ideali.

NB: per riprodurre i risultati in autonomia ricordarsi di disattivare il 'rumore' sulle misure della pressione.

Un gas ideale è caratterizzato da pressione (p), temperatura (T) e volume (V).
Fissare a rotazione e vedere come variano gli altri due, riportare i valori in un grafico per studiare la dipendenza.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Pressione Costante


```python
p=11620 #kPa
```

```python
T_mis_C=np.array([0,10,25,50,75,100]) #°C
L_mis_nm=np.array([5,5.2,5.4,5.9,6.4,6.8])
A_mis_nm2=5*5
V_mis_nm3=A_mis_nm2*L_mis_nm
```

```python
plt.plot(T_mis_C,V_mis_nm3,linestyle='None',marker='o')
plt.xlabel('Temperatura (°C)')
plt.ylabel('Volume (nm³)')
plt.show()
```

```python
m=np.polyfit(T_mis_C,V_mis_nm3,1)
m
```

```python
T=np.linspace(-273.15,100,100)
V=m[0]*T+m[1]
```

```python
plt.plot(T,V,linestyle='--',color='red')
plt.plot(T_mis_C,V_mis_nm3,linestyle='None',marker='o')
plt.xlabel('Temperatura (°C)')
plt.ylabel('Volume (nm³)')
plt.xlim(-273.15,100)
plt.ylim(0,175)
plt.show()
```

### Trasformazione nel Diagramma di Clapeyron (p vs V)

```python
p_mis=p*np.ones(len(V_mis_nm3))
```

```python
from matplotlib import color_sequences
plt.plot(V_mis_nm3,p_mis,linestyle='--',color='red')
plt.plot([V_mis_nm3[0],V_mis_nm3[-1]],p_mis[0:2],linestyle='None',marker='o',color='red')
plt.xlabel('Volume (nm³)')
plt.ylabel('Pressione (kPa)')
```

```python
T_mis_C=np.array([20,50,74,91,120])
p_mis_kPa =np.array([4041,4459,4790,5030,5424])
```

```python
plt.plot(T_mis_C,p_mis_kPa,linestyle='None',marker='o')
plt.xlabel('Temperatura (°C)')
plt.ylabel('Pressione (kPa)')
```

```python
m=np.polyfit(T_mis_C,p_mis_kPa,1)
m
```

```python
T=np.linspace(-273.15,150,100)
p=m[0]*T+m[1]
```

```python
plt.plot(p,linestyle='--',color='red')
plt.plot(T_mis_C,p_mis_kPa,linestyle='None',marker='o')
plt.xlabel('Temperatura (°C)')
plt.ylabel('Pressione (kPa)')
plt.xlim(-273.15,150)
plt.ylim(0,6000)
plt.show()
```
